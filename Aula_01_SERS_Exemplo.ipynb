 Motor de Indução - Consumo e Demanda**

inseção dos dados de placa do motor

PU-----> potêcia útil (mecanica), em watts (W)

n------> rendimento ----> n= PU/P

fp -----> fator de potência----> fp= P/S

```python
# Dados de entrada
# Potência útil (PU)
# Rendimento (n)
# Fator de potencia (fp)
PU = int(input('Insira o valor da potência útil em Watts: '))
n = float(input('Insira o valor do rendimento (de 0 a 1): '))
fp = float(input('Insira o valor do fator de potencia (de 0 a 1):'))
```

```python
PU, n, fp
```

2. calculo das potencias ativa (P), aparente (S) e reativa (Q)


```python
# Cálculo da potência ativa (P) em watts
# n = PU / P
P = PU / n
# potencia aparente (S), VA
# fp = P / S
S = P / fp
#potencia reativa (Q), em VAr
# para esponeciação, usar: (**)
Q = (S**2 - P**2)
```

```python
print(f'A potência ativa do motor será de: {int(P)} W')
```

3. aprensentação dos resultados
f-strings -----> permite utilizar variaveis dentroo da string principal

```python
print(f'a potencia ativa sera de {int(P)}W ') #somente valore inteiro
print(f'a potencia aparente sera de {int(S)}Va')
print(f'a potencia reativa sera de {int(Q)}W')
```

```python
# Fator de potência (FP)
FP = float(input('Insira o fator de potência (de 0 a 1): '))
```

```python
# Cálculo da potência total (Potência Aparente --> S)
# FP = P / S
S = P / FP
```

```python
print(f'A potência aparente (total) para o motor é de: {int(S)}')
```
